```python
df4 = pd.read_csv('flow_csv/slowloris.pcap_Flow.csv')
df4.head()
```

```python
df4 = df4[df4['Src IP'] == '10.5.19.145']  
```

```python
df4.drop(columns=['Flow ID', 'Src IP', 'Src Port', 'Dst IP', 'Dst Port', 'Protocol', 'Timestamp'], inplace=True)
```

```python
# Gera métricas por coluna e salva em CSV
metricas_df_slowloris = df4.describe(include="all").T

metricas_df_slowloris["dtype"] = df4.dtypes.astype(str)
metricas_df_slowloris["non_null"] = df4.notna().sum()
metricas_df_slowloris["null_count"] = df4.isna().sum()
metricas_df_slowloris["null_pct"] = (df4.isna().mean() * 100).round(4)
metricas_df_slowloris["n_unique"] = df4.nunique(dropna=True)

# Reorganiza colunas para facilitar leitura
ordem = ["mean", "std", "min", "25%", "50%", "75%", "max", ]
metricas_df_slowloris = metricas_df_slowloris[[c for c in ordem if c in metricas_df_slowloris.columns]]

# Salva arquivo
arquivo_saida = "simulacao.csv"
metricas_df_slowloris.to_csv(arquivo_saida, encoding="utf-8-sig")

print(f"CSV gerado com sucesso: {arquivo_saida}")
metricas_df_slowloris.head()
```

```python
import os
import seaborn as sns

import matplotlib.pyplot as plt

# Colunas numéricas (exclui constantes)
numeric_cols = df4.select_dtypes(include=["number"]).columns.tolist()
numeric_cols = [c for c in numeric_cols if df4[c].nunique(dropna=True) > 1]

# Gera um PNG por coluna: histograma com KDE + boxplot
for col in numeric_cols:
    vals = df4[col].dropna()
    fig, axes = plt.subplots(2, 1, figsize=(8, 5), gridspec_kw={"height_ratios": [4, 1]})
    sns.histplot(vals, kde=True, ax=axes[0], color="C0")
    axes[0].set_title(f"{col} — histograma + KDE")
    axes[0].set_ylabel("contagem")
    sns.boxplot(x=vals, ax=axes[1], color="C1")
    axes[1].set_title(f"{col} — boxplot")
    axes[1].set_xlabel("")
    plt.tight_layout()
    plt.show()
    plt.close(fig)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Lê os dois CSVs
simulacao = pd.read_csv("simulacao.csv")
metricas = pd.read_csv("metricas_df_slowloris.csv")

# Ajusta índice (caso tenha sido salvo como "Unnamed: 0")
def preparar_df(df):
    df = df.copy()
    unnamed_cols = [c for c in df.columns if c.lower().startswith("unnamed")]
    if unnamed_cols:
        df = df.rename(columns={unnamed_cols[0]: "feature"}).set_index("feature")
    return df

simulacao = preparar_df(simulacao)
metricas = preparar_df(metricas)

# Mantém apenas colunas numéricas em comum
simulacao_num = simulacao.apply(pd.to_numeric, errors="coerce")
metricas_num = metricas.apply(pd.to_numeric, errors="coerce")

common_cols = simulacao_num.columns.intersection(metricas_num.columns)
common_idx = simulacao_num.index.intersection(metricas_num.index)

sim_alinhado = simulacao_num.loc[common_idx, common_cols]
met_alinhado = metricas_num.loc[common_idx, common_cols]

# Diferença entre os dois (simulacao - metricas)
diff = sim_alinhado - met_alinhado

# 1) Heatmap das diferenças por linha/coluna
plt.figure(figsize=(12, max(4, len(common_idx) * 0.25)))
sns.heatmap(diff, cmap="coolwarm", center=0, linewidths=0.3, linecolor="white")
plt.title("Diferenças por feature e métrica (simulacao - metricas_df_slowloris)")
plt.xlabel("Métricas")
plt.ylabel("Features")
plt.tight_layout()
plt.show()

# 2) Média da diferença absoluta por coluna
media_diff_abs = diff.abs().mean(axis=0).sort_values(ascending=False)

plt.figure(figsize=(10, 4))
sns.barplot(x=media_diff_abs.index, y=media_diff_abs.values, color="C0")
plt.title("Média da diferença absoluta por métrica")
plt.xlabel("Métricas")
plt.ylabel("Diferença absoluta média")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()

# (Opcional) tabela resumida das maiores diferenças
resumo = pd.DataFrame({
    "mean_abs_diff": diff.abs().mean(),
    "max_abs_diff": diff.abs().max(),
    "mean_diff": diff.mean()
}).sort_values("mean_abs_diff", ascending=False)

resumo.head(10)
```
